```python
import pandas as pd
import numpy as np

df = pd.read_csv("adult.csv")

print(df.head())
```

```python
print("Task 1")
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
age_array = df["age"].to_numpy()
hours_array = df["hours.per.week"].to_numpy()

print("Age NumPy Array:")
print(age_array[:10])

print("\nHours-per-week NumPy Array:")
print(hours_array[:10])

print("\nData types:")
print("Age:", age_array.dtype)
print("Hours-per-week:", hours_array.dtype)
```

```python

print("First age:", age_array[0])
print("Second age:", age_array[1])
print("First working hours:", hours_array[0])
print("Second working hours:", hours_array[1])
```

```python
basic_slice = age_array[0:10]

print("Basic Slicing - first 10 ages:")
print(basic_slice)
```

```python
fancy_index = age_array[[0, 5, 10, 15, 20]]

print("Fancy Indexing:")
print(fancy_index)
```

```python
print("Task 2")
age_array = df["age"].to_numpy()

hours_array = df["hours.per.week"].to_numpy()

scaled_value = age_array * (hours_array / 40)

print("First 10 age values:")
print(age_array[:10])

print("\nFirst 10 hours-per-week values:")
print(hours_array[:10])

print("\nFirst 10 derived values:")
print(scaled_value[:10])
```

```python
loop_result = []

for i in range(len(age_array)):
    loop_result.append(age_array[i] * (hours_array[i] / 40))

print("First 10 loop results:")
print(loop_result[:10])
```

```python
print(np.allclose(scaled_value, loop_result))
%timeit age_array * (hours_array / 40)
%timeit [age_array[i] * (hours_array[i] / 40) for i in range(len(age_array))]
```

```python
print("task 3")
filtered_data = df[
    (df["age"] > 40) &
    (df["education"] == "Bachelors") &
    (df["hours.per.week"] > 40)
]

print("Filtered Dataset:")
print(filtered_data.head(10))

print("\nNumber of matching records:", len(filtered_data))
```

```python
print(filtered_data[
    ["age", "education", "hours.per.week", "income"]
].head(10))
```

```python
print("Age condition:", (filtered_data["age"] > 40).all())
print("Education condition:", (filtered_data["education"] == "Bachelors").all())
print("Hours condition:", (filtered_data["hours.per.week"] > 40).all())
```

```python
import os
os.makedirs("Output_Files", exist_ok=True)
filtered_data.to_csv(
    "Output_Files/filtered_dataset.csv",
    index=False
)
print("Filtered dataset saved successfully.")
```

```python
print("task 4")
grouped_data = df.groupby(
    "education",
    as_index=False
)["hours.per.week"].mean()

print("GroupBy Result:")
print(grouped_data)
```

```python
grouped_data.to_csv(
    "Output_Files/groupby_result.csv",
    index=False
)

grouped_data.to_excel(
    "Output_Files/groupby_result.xlsx",
    index=False
)

print("CSV and XLSX files saved successfully.")
```

```python
print("task 5")
income_counts = df["income"].value_counts()
print("Income Value Counts:")
print(income_counts)
```

```python
income_percentage = df["income"].value_counts(normalize=True) * 100
print("Income Percentage Distribution:")
print(income_percentage.round(2))
```

```python
majority_count = income_counts.max()
minority_count = income_counts.min()
imbalance_ratio = majority_count / minority_count
print("Majority class count:", majority_count)
print("Minority class count:", minority_count)
print("Imbalance ratio:", round(imbalance_ratio, 2))
```

```python
dominant_class = income_counts.idxmax()
dominant_percentage = income_percentage.max()

print("Dominant value:", dominant_class)
print("Dominant percentage:", round(dominant_percentage, 2), "%")
```
